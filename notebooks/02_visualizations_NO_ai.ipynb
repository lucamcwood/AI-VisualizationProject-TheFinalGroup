```python
import pandas as pd
import seaborn as sns
import os
import matplotlib.pyplot as plt

from google.colab import drive
drive.mount("/content/drive")

# Load the data
data_path = "/content/drive/My Drive/VIS4AI/results/run1/r1-distill.jsonl"

df = pd.read_json(data_path, lines=True)
df
```

```python
# New features
df['words_in'] = df['question'].apply(lambda x: len(x.split()))
df['words_out'] = df['output'].apply(lambda x: len(x.split()))
df
```

```python
sns.set(style="whitegrid")

def plot_scatter(df, x_col, y_col, hue, title, figsize=(10, 6)):
    plt.figure(figsize=(10, 6))
    sns.scatterplot(data=df, x=x_col, y=y_col, hue=hue)
    plt.title(title)
    plt.xlabel(x_col)
    plt.ylabel(y_col)
    plt.show()
    
plot_scatter(df, 'correct', 'n_tokens', 'finished', 'Scatter Plot of X vs Y')  # Replace with actual column names
```
